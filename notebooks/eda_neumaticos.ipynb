```python
# Imports necesarios
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import requests
import json
from datetime import datetime, date
import warnings
warnings.filterwarnings('ignore')

# Configuración de visualización
plt.style.use('seaborn-v0_8')
sns.set_palette('husl')
plt.rcParams['figure.figsize'] = (12, 8)
```

## 1. Extracción de Datos desde API

```python
# Configuración API
API_BASE_URL = "http://localhost:8001/api/v1"
TRAINING_DATA_ENDPOINT = f"{API_BASE_URL}/ml/training-data"

# Función para obtener datos
def fetch_training_data(params=None):
    """Obtiene datos de entrenamiento desde la API."""
    try:
        response = requests.get(TRAINING_DATA_ENDPOINT, params=params)
        response.raise_for_status()
        return response.json()
    except requests.exceptions.RequestException as e:
        print(f"Error al obtener datos: {e}")
        return None

# Obtener datos completos
print("Obteniendo datos de entrenamiento...")
raw_data = fetch_training_data()

if raw_data:
    print(f"✅ Datos obtenidos: {raw_data['total_records']} registros")
    print(f"📊 Estadísticas: {raw_data['summary_stats']}")
else:
    print("❌ No se pudieron obtener los datos")
```

## 2. Preparación del Dataset

```python
# Convertir a DataFrame
if raw_data and raw_data['data_points']:
    df = pd.DataFrame(raw_data['data_points'])
    
    # Conversión de tipos
    date_columns = ['fecha_compra', 'fecha_fabricacion', 'fecha_desecho', 'fecha_ultima_prediccion']
    for col in date_columns:
        if col in df.columns:
            df[col] = pd.to_datetime(df[col])
    
    # Información básica del dataset
    print(f"📋 Shape del dataset: {df.shape}")
    print(f"📋 Columnas: {list(df.columns)}")
    print(f"📋 Tipos de datos:")
    print(df.dtypes)
else:
    print("❌ No hay datos para procesar")
    df = pd.DataFrame()
```
